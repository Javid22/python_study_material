```python
import pandas as pd
from functools import reduce

orders_dataset = pd.read_csv("files/olist_orders_dataset.csv")
order_items_dataset = pd.read_csv("files/olist_order_items_dataset.csv")
order_review_dataset = pd.read_csv('files/olist_order_reviews_dataset.csv')
order_payments_dataset = pd.read_csv('files/olist_order_payments_dataset.csv')
order_products_dataset = pd.read_csv('files/olist_products_dataset.csv')
order_sellers_dataset = pd.read_csv('files/olist_sellers_dataset.csv')
order_customer_dataset = pd.read_csv('files/olist_customers_dataset.csv')
order_geolocation_dataset = pd.read_csv('files/olist_geolocation_dataset.csv')

dfs = [
    orders_dataset,
    order_items_dataset,
    order_payments_dataset,
    order_review_dataset
]

merged_df = reduce(lambda left, right: pd.merge(left, right, on="order_id", how="inner"), dfs)


```

```python
merged_df = pd.merge(merged_df, order_products_dataset, on="product_id", how="left")
merged_df = pd.merge(merged_df, order_sellers_dataset, right_on="seller_id", left_on="seller_id",how="left")
merged_df = pd.merge(merged_df, order_customer_dataset, on="customer_id", how="left")
```

```python
merged_df.columns
merged_df.
```

```python
geo = (
    order_geolocation_dataset
    .groupby("geolocation_zip_code_prefix", as_index=False)
    .agg({
        "geolocation_lat": "mean",
        "geolocation_lng": "mean"
    })
)

merged_df = merged_df.merge(geo, left_on="customer_zip_code_prefix", right_on="geolocation_zip_code_prefix", how="left")
```
