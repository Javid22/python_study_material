## OOPS

class , object, Inheritance, encapsulation, Polymorphism, abstractmethod

```python
from abc import ABC, abstractmethod

class Employee(ABC):  # Class + Abstract class
    def __init__(self, name, designation, salary): ## constructure
        self.name = name
        self.designation = designation
        self.__salary = salary ## encapsulation

    def get_name(self): # Method
        return self.name

    def get_designation(self): # Method
        return self.designation

    def get_salary(self):  # Method to access private salary
        return self.__salary

    def check_salary(self): # Method
        if self.__salary > 0:
            return True
        return False

    @abstractmethod
    def salary_increment_cal(self):
        pass

class Team(Employee): ## Inheritance
    def get_ticket(self):  # Method
        return "your ticket number is 101"

    def salary_increment_cal(self): ## Polymorphism
        return self.get_salary() + (self.get_salary() * 0.10)

    def work(self): ## Polymorphism
        return f"As {self.designation} your job is to clear assigned ticket"

class Manager(Employee): ## Inheritance
    def assign_ticket(self):  # Method
        return "tell me which ticket you want to assign"

    def work(self): ## Polymorphism
            return f"As {self.designation} your job is to assign task to team member"

    def salary_increment_cal(self): ## Polymorphism
        return self.get_salary() + (self.get_salary() * 0.20)
```
