```python
concated_data = pd.concat([sales_trans_2017, sales_trans_2018, sales_trans_2019])
```

```python
concated_data['Voucher'] = concated_data['Voucher'].replace(r'^Sal:', '', regex=True)
```

```python
concated_data.dropna(axis= 'columns')
```

```python
concated_data.drop(['Gross', 'Disc', 'Voucher Amount'], axis=1, inplace=True)
```

```python
concated_data.drop_duplicates()
```

```python
result = concated_data.dropna(subset=['Party', 'Product', 'Voucher', 'Date'])
```

```python
result.isnull().sum()
```

```python
print(result['Rate'].unique())
```

```python
result['Qty'] = result['Qty'].replace(",", "", regex=True)
result['Qty'] = pd.to_numeric(result['Qty'], errors='coerce').astype('Float64')
```

```python
result.loc[:, 'Voucher']
```

```python
result['Date'] = pd.to_datetime(result['Date'].replace(", ", "", regex=True), errors='coerce').astype('Float64')
result['Voucher'] = pd.to_numeric(result[ 'Voucher'].replace(",", "", regex=True), errors='coerce').astype('Float64')
```

```python
result['Rate'] = result['Rate'].replace(",", "", regex=True)
result['Rate'] = pd.to_numeric(result['Rate'], errors='coerce').astype('Float64')
```
