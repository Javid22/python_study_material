Instructions:
1. Create a list of random words (like “python”, “machine”, “learning”, “bridge”).
2. Randomly select one word for the game.
3. Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
4. Keep track of wrong guesses (max 6 attempts).
5. End the game when the player either guesses the full word or uses all chances.

```python
import random
print("*-----------------  WELCOME TO HANGMAN -----------------*")
print("Tip: The word will be based on a state in India. Players have a maximum of 6 incorrect guesses. They can guess any character that may appear in the word.")

rand_words = ["mumbai", "chennai", "gujarat", "delhi", "bangalore", "hyderabad", "kolkata", "pune", "jaipur", "lucknow"]
totalattempt = 6
selectedword = random.choice(rand_words)
hash = ["_"] * len(selectedword)


print("GUESS THE FIRST CHARACTER ", " ".join(hash))
word = input("Enter First Character").lower()[0]
while totalattempt > 0:
    if word in selectedword:
        print("you have guessed correct character")
        for i, v in enumerate(selectedword):
            if v == word:
                hash[i] = word
        if selectedword == "".join(hash):
            print("Congratulations! You have guessed the word:", selectedword)
            break
        print("guess the next word", " ".join(hash))
        word = input("Enter Next Character").lower()[0]
        if word in hash:
            print("you have already guessed this character, please guess another character")
            word = input("Enter Next Character").lower()[0]
    else:
        totalattempt -= 1
        print("wrong guess, remaing attempt", totalattempt)
        print("guess the next word", " ".join(hash))
        word = input("Enter Next Character").lower()[0]


print("Game Over, you have exhausted all attempts. The word was:", selectedword)
```

```python
import random
print("Welcome to Rock, Paper, Scissors Game")
user_input = input("Enter Your input as (Rock/Paper/Scissors)")

data = ["Rock", "Paper", "Scissors"]
while True:
    try:
        sys_input = random.choice(data)
        user_input = user_input.capitalize()
        print(sys_input, user_input)
        if (user_input not in data):
            print("please enter valid input between Rock, Paper, Scissors ")
            user_input = input("Enter Your input as (Rock/Paper/Scissors)")
        if (user_input in data):
            if (sys_input == user_input):
                print("DRAW - RETRY")
            elif ((sys_input == "Rock" and user_input == "Scissors") or (sys_input == "Paper" and user_input == "Rock") or (sys_input == "Scissors" and user_input == "Paper")):
                print("YOU WIN")
            elif ((user_input == "Rock" and sys_input == "Scissors") or (user_input == "Paper" and sys_input == "Rock") or (user_input == "Scissors" and sys_input == "Paper")):
                print("YOU WIN")
            
            play_again = input("\nDo you want to play again? (yes/no): ").lower()

            if play_again != "yes":
                print("Thanks for playing!")
                break
            if play_again.lower() == "yes":
                user_input = input("Enter Your input as (Rock/Paper/Scissors)")
    except:
        print("OOPS! please try again between Rock, Paper, Scissors ")
    
```

Your program should:

Remove extra spaces.
Check whether the ID starts with "EMP-".
Ensure the remaining part contains exactly 4 digits.
Display whether the ID is valid.

```python
empId = input("Enter Employee Id")
isValidEmpId = empId.strip().upper().startswith("EMP-") and len(empId) == 8 and empId[4:].isdigit()

if isValidEmpId:
    print("Valid Employee Id")
else:
    print("Invalid Employee Id. Please enter a valid Employee Id in the format 'EMP-XXXX' where XXXX are digits.")
```

```python
password = input("Enter Password")

if len(password) < 8:
    print("Password must be at least 8 characters long.")
elif not any(char.isdigit() for char in password):
    print("Password must contain at least one digit.")
elif not any(char.isupper() for char in password):
    print("Password must contain at least one uppercase letter.")
elif not any(char.islower() for char in password):
    print("Password must contain at least one lower letter.")
else:
    print("Password is valid.")
```
