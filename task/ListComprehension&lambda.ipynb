Online Shopping Order Analysis

Part 1: List Comprehension (2 Tasks)  
Task 1 -  Create a list containing only the customer names whose orders are Delivered.

```python
orders = [
    {"customer": "Alice", "amount": 1200, "status": "Delivered"},
    {"customer": "Bob", "amount": 450, "status": "Cancelled"},
    {"customer": "Charlie", "amount": 1800, "status": "Delivered"},
    {"customer": "David", "amount": 750, "status": "Delivered"},
    {"customer": "Eva", "amount": 2500, "status": "Delivered"},
    {"customer": "Frank", "amount": 300, "status": "Cancelled"}
]
```

```python
order_delivered_name = []
for order in orders:
    if order['status'] == "Delivered":
        order_delivered_name.append(order['customer'])
print(order_delivered_name)
```

Task 2 - Create a list of order amounts after applying a 10% discount only to delivered orders.

```python
amount_after_discount = []
for order in orders:
    if order['status'] == "Delivered":
        discount = order['amount'] - (order['amount'] * 10 / 100)
        amount_after_discount.append(discount)
print(amount_after_discount)
```

Part 2: Lambda + map() (2 Tasks)

Task 3 - 
Using map() and a lambda function, create a list where every order amount has 18% GST added.

```python
amount_after_gst = list(map(lambda order: order['amount'] + (order['amount'] * 18 / 100), orders))
print(amount_after_gst)
```

Task 4 - Using map(), convert every customer name to uppercase.

```python
customer_name = list(map(lambda order: order['customer'].upper(), orders))
customer_name
```

Part 3: Lambda + filter() (2 Tasks)

Task 5 - Using filter(), get all orders whose amount is greater than 1000.

```python
eligible_amount = list(filter(lambda order: order['amount']>= 1000, orders))
eligible_amount
```

Task 6
Using filter(), retrieve only the Cancelled orders.

```python
cancelled_order = list(filter(lambda order: order['status'] == 'Cancelled', orders))
cancelled_order
```

Part 4: Lambda + reduce() (2 Tasks)

Task 7
Find the total revenue (sum of all order amounts).

```python
from functools import reduce
total_amount = reduce(lambda x,y: x+y, map(lambda order: order['amount'], orders))
total_amount
```

Task 8 - Find the customer who placed the highest-value order

```python
from functools import reduce

higer_order = reduce(lambda x,y: x if x > y else y, map(lambda order: order['amount'], orders))
higer_order
```

task 9 - Create a dictionary counting orders by status.

```python
from functools import reduce

result = reduce(lambda x,y: {
        **x,
        y["status"]: x.get(y['status'], 0) + 1
    }, orders, {})
result
# reduce(function, iterable, initial_value)
```

```python
from functools import reduce

result = reduce(lambda x,y: {
        **x,
        y["status"]: x.get(y['status'], 0) + y['amount']
    }, orders, {})
result
```
