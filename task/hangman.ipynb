    1. Create a list of random words (like “python”, “machine”, “learning”, “bridge”).
    2. Randomly select one word for the game.
    3. Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
    4. Keep track of wrong guesses (max 6 attempts).
    5. End the game when the player either guesses the full word or uses all chances.

```python
import random
print("WELCOME TO HANGMAN")
print("Tip: Word will be based on stated in India. max 6 attempt of wrong guess. player can say any character within that word.")

rand_words = ["delhi", "mumbai", "chennai", "banglore", "gujarat"]
totalattempt = 6
selectedword = random.choice(rand_words)
print(selectedword)
hash = ""
for i in selectedword:
    hash += " _"

print("GUESS THE FIRST CHARACTER ", hash)
word = input("Enter First Character")

if word in selectedword:
    print("you have guessed correct character")
    hash = ""
    for i in selectedword:
        if i == word:
            hash += i
        else:
            hash += " _"
    print("guess the next word", hash)
    word = input("Enter First Character")
else :
    print("wrong guess remaining count", totalattempt - 1)
```
