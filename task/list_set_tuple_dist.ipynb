```python
list_val = [10,20,30]

print(list_val)
list_val.append(40)
list_val.pop()
list_val.remove(30)
val2 = list_val.copy()
count = list_val.count(10)
list_val.extend([50,60,70])
sorting = [10,20, 60, 40, 30] #["a", "b", "e", "k", "l"]
sorting.sort()  # Sorts the list in place
sorting.reverse()
print(sorting)
print(list_val)
print(count)
```

2. SET

```python
set_val = set()
print(type(set_val))
set_val = {10, 20, 30, 40, 50}
set_val.add(60)
set_val.update([70, 80, 90])
set_val.discard(20)
set_val = set_val.union({100, 200, 300})
set_val.pop()
set_val.intersection({10, 20, 30, 40, 50})
set_val.issuperset({10, 20, 30})
print(set_val)
```

3. Tuple

```python
tuple_val = (10, 20, 30, 40, 50)
print(tuple_val.count(10))
print(tuple_val.index(40))
```

4. Dictionary

```python
dict_val = {"name": "John", "age": 30, "city": "New York"}
dict_val["age"] = 31
print(dict_val)
dict_val.fromkeys(["name", "city"])
print(dict_val.values())
dict_val.update({"gender": "Male"})
dict_val.popitem()
dict_val.pop("age")
print(dict_val.items())
dict_val.setdefault("age", 30)
dict_val.update({"age": 22})
dict_val.clear()
print(dict_val)
```
