```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier, RandomForestClassifier
from sklearn.feature_selection import SelectFromModel, RFE
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
import warnings
warnings.filterwarnings('ignore')
```

```python
card_dataset = pd.read_csv("classification(discreate)/DecisionTree/credit_card_clean.csv")
card_dataset
```

```python
label_encoder = LabelEncoder()
card_dataset['SEX'] = label_encoder.fit_transform(card_dataset['SEX'])
card_dataset['EDUCATION'] = label_encoder.fit_transform(card_dataset['EDUCATION'])
card_dataset['MARRIAGE'] = label_encoder.fit_transform(card_dataset['MARRIAGE'])
```

```python
card_dataset
```

```python
X= card_dataset.drop('DEFAULT', axis=1)
Y= card_dataset['DEFAULT']
X_train, X_test, y_train, y_test = train_test_split(X,Y, random_state=123, test_size=0.20)
```

```python
ada_clasifier = AdaBoostClassifier()
```

```python
select_from_model = SelectFromModel(estimator=AdaBoostClassifier(), max_features=7)
```

```python
select_from_model.fit_transform(X_train, y_train)
```

```python
select_from_model.get_support()
```

```python
X.columns[select_from_model.get_support()]
```

```python
X_train_select_from_model = select_from_model.transform(X_train)
X_test_select_from_model = select_from_model.transform(X_test)
```

```python
def runRandomForestClassifier(X_train,X_test,y_train,y_test):
    rf_model = RandomForestClassifier(max_depth=7,random_state=123)
    rf_model.fit(X_train,y_train)
    
    y_pred = rf_model.predict(X_test)
    
    print("Accuracy Score:",round(accuracy_score(y_test,y_pred),4))
    print("Precision Score:",round(precision_score(y_test,y_pred),4))
    print("Recall Score",round(recall_score(y_test,y_pred),4))
    print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred))
```

```python
%%time
runRandomForestClassifier(X_train,X_test,y_train,y_test)
```

```python
%%time
runRandomForestClassifier(X_train_select_from_model,X_test_select_from_model,y_train,y_test)
```

```python
## Second Techniq RFE(,Recursive Feature Elimination)
```

```python
rfe_model = RFE(estimator = RandomForestClassifier(),n_features_to_select=5)
```

```python
rfe_model.fit_transform(X_train,y_train)
```

```python
X.columns[rfe_model.get_support()]
```

```python
X_train_select_from_model = rfe_model.transform(X_train)
X_test_select_from_model = rfe_model.transform(X_test)
```

```python
%%time
runRandomForestClassifier(X_train,X_test,y_train,y_test)
```

```python
%%time
runRandomForestClassifier(X_train_select_from_model,X_test_select_from_model,y_train,y_test)
```

```python
### 3. Take the Important Features from GradientBoosting and train it within RandomForest Algorithm
```

```python
select_from_model = SelectFromModel(estimator=GradientBoostingClassifier())
select_from_model.fit_transform(X_train,y_train)
```

```python
X.columns[select_from_model.get_support()]
```

```python
x_train_pred_data = select_from_model.transform(X_train)
x_test_pred_data = select_from_model.transform(X_test)
```

```python
def runRandomForestClassifier(X_train,X_test,y_train,y_test):
    rf_model = RandomForestClassifier(max_depth=7,random_state=123)
    rf_model.fit(X_train,y_train)
    
    y_pred = rf_model.predict(X_test)
    
    print("Accuracy Score:",round(accuracy_score(y_test,y_pred),4))
    print("Precision Score:",round(precision_score(y_test,y_pred),4))
    print("Recall Score",round(recall_score(y_test,y_pred),4))
    print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred))
```

```python
runRandomForestClassifier(x_train_pred_data, x_test_pred_data, y_train, y_test)
```

```python
a### 4. How to choose the Optimal Number of Features to train my Model?
```

```python
for i in range(1,31):
    rfe_model = RFE(estimator = GradientBoostingClassifier(),n_features_to_select=i)
    rfe_model.fit_transform(X_train, y_train)
    x_train_pred_data = rfe_model.transform(X_train)
    x_test_pred_data = rfe_model.transform(X_test)
    print("Selected Features:",i)
    runRandomForestClassifier(x_train_pred_data,x_test_pred_data,y_train,y_test)
    print("**************************")
```
